**Цель семинара:** Понять, как работает градиентный спуск, и реализовать его с нуля на Python для простой функции, а затем для задачи линейной регрессии.

**Градиентный спуск** — это итерационный алгоритм оптимизации для нахождения локального минимума функции. Идея заключается в том, чтобы двигаться в направлении, противоположном градиенту функции, так как градиент указывает направление наискорейшего роста.

Представьте, что вы стоите на склоне горы в густом тумане и хотите спуститься вниз. Куда сделать шаг? Самый очевидный ответ — в сторону самого крутого спуска. Градиент — это вектор, который показывает направление самого крутого подъёма. Соответственно, чтобы спускаться, нам нужно двигаться в обратном направлении (анти-градиент).

Шаг алгоритма выглядит так:
$$ w_{new} = w_{current} - \alpha \cdot \nabla f(w_{current}) $$
где:
- $w_{new}$ — новая точка (наши новые параметры).
- $w_{current}$ — текущая точка.
- $\alpha$ (альфа) — скорость обучения (learning rate), которая контролирует размер шага.
- $\nabla f(w_{current})$ — градиент функции в текущей точке.

### Финальный штрих: обучение модели и визуализация

Теперь мы соберем всё вместе. Код ниже уже написан. Он использует вашу функцию `calculate_mse_gradient` для обучения модели линейной регрессии.

Мы сделаем два вида визуализации:
1.  Как меняется положение регрессионной прямой в процессе обучения.
2.  Как параметры ($w_0, w_1$) движутся по "карте высот" (контурному графику) функции потерь MSE.

```python
# Гиперпараметры
learning_rate = 0.1
num_iterations = 100

# Начальная инициализация весов
w0 = 0
w1 = 0

# Хранилище для истории весов и потерь
w_history = [(w0, w1)]
loss_history = []

# Цикл обучения
for i in range(num_iterations):
    # Считаем градиент
    grad_w0, grad_w1 = calculate_mse_gradient(X, y, w0, w1)
    
    # Обновляем веса
    w0 = w0 - learning_rate * grad_w0
    w1 = w1 - learning_rate * grad_w1
    
    # Сохраняем историю
    w_history.append((w0, w1))
    loss = np.mean((y - (w1 * X + w0))**2)
    loss_history.append(loss)

print(f"Обучение завершено. Финальные параметры: w0 = {w0:.2f}, w1 = {w1:.2f}")

# Визуализация 1: Регрессионная прямая
plt.figure(figsize=(10, 6))
plt.scatter(X, y, alpha=0.6, label='Данные')
x_line = np.array([0, 2])
# Покажем несколько этапов обучения
for i in [0, 1, 5, 10, -1]:
    w0_i, w1_i = w_history[i]
    y_line = w1_i * x_line + w0_i
    label = f'Итерация {i}' if i != -1 else 'Финальная'
    plt.plot(x_line, y_line, label=label, alpha=0.8)
plt.title('Процесс обучения линейной регрессии')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.show()

# Визуализация 2: Поверхность потерь
w0_vals = np.linspace(w0 - 2, w0 + 2, 100)
w1_vals = np.linspace(w1 - 2, w1 + 2, 100)
W0, W1 = np.meshgrid(w0_vals, w1_vals)
Loss = np.zeros_like(W0)

for i in range(W0.shape[0]):
    for j in range(W0.shape[1]):
        y_pred = W1[i, j] * X + W0[i, j]
        Loss[i, j] = np.mean((y - y_pred)**2)

plt.figure(figsize=(10, 8))
plt.contourf(W0, W1, Loss, levels=np.logspace(0, 2, 20), cmap='viridis')
plt.colorbar(label='MSE (log scale)')
w_path = np.array(w_history)
plt.plot(w_path[:, 0], w_path[:, 1], 'r-o', label='Путь градиентного спуска')
plt.title('Спуск по поверхности потерь MSE')
plt.xlabel('$w_0$')
plt.ylabel('$w_1$')
plt.legend()
plt.show()
```
