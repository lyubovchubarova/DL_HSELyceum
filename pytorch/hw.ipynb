# СПОЙЛЕРЫ!

Рекомендуемый план действий:

* Адаптируйте логистическую регрессию из задания семинара для классификации одной буквы против остальных (например, A против остальных)
* Обобщите ее до многоклассовой логистической регрессии.
  - Либо попробуйте вспомнить лекцию 0, либо погуглите.
  - Вместо вектора весов вам придется использовать матрицу (идентификатор_признака x идентификатор_класса)
  - Softmax (экспонента над суммой экспонент) может быть реализован вручную или как `nn.Softmax` (слой) или `F.softmax` (функция)
  - Вероятно, лучше использовать СТОХАСТИЧЕСКИЙ градиентный спуск (мини-батчи) для большей скорости
    - Вы также можете попробовать momentum/rmsprop/adam и т.д.
    - в этом случае набор данных, вероятно, должен быть перемешан (или использовать случайные подвыборки на каждой итерации)
* Добавьте скрытый слой. Теперь ваша логистическая регрессия использует скрытые нейроны вместо входов.
  - Скрытый слой использует ту же математику, что и выходной слой (бывшая логистическая регрессия), но использует некоторую нелинейность (например, сигмоиду) вместо softmax
  - Вам нужно обучать оба слоя, а не только выходной слой :)
  - 50 скрытых нейронов и сигмоидная нелинейность подойдут для начала. Много способов улучшить.
  - В идеальном случае это составляет 2 `torch.matmul`, 1 softmax и 1 ReLU/сигмоиду
  - __Убедитесь, что эта нейронная сеть работает лучше, чем логистическая регрессия!__

* Теперь самое время попробовать улучшить сеть. Рассмотрите слои (размер, количество нейронов), нелинейности, методы оптимизации, инициализацию — что угодно, но, пожалуйста, избегайте сверток пока.

* Если что-то кажется неправильным, попробуйте пройти один шаг обучения и вывести все, что вы вычисляете.
* Если вы видите NaN посреди оптимизации, вы можете оценить $\log P(y \mid x)$ как `F.log_softmax(layer_before_softmax)`.
